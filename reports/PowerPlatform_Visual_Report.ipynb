## 1 · Overview
The platform adoption figures describe **FY2025 Q3**, not a 2026 measurement. The UK counts use two separate six-month vacancy samples. Read each indicator with its period and source limitation.


```python
# @title Render the visual overview
adoption = frames["Adoption"]
hiring = frames["UKHiring"].sort_values("Period")
mau = adoption.loc[adoption.Unit.eq("Million users"), "Value"].iloc[0]
yoy = adoption.loc[adoption.Unit.eq("Percent YoY"), "Value"].iloc[0]
prior, latest = hiring.Vacancies.iloc[0], hiring.Vacancies.iloc[-1]
growth = (latest / prior - 1) * 100
share_delta = hiring.SharePct.iloc[-1] - hiring.SharePct.iloc[0]
fig = plt.figure(figsize=(13,7.4))
gs = fig.add_gridspec(3,4, height_ratios=[0.75,1.8,0.35], hspace=0.6, wspace=0.55)
for i, (value, label) in enumerate([(f"{mau:.0f}M","Reported monthly users\nFY2025 Q3"),
    (f"{yoy:.0f}%","Reported user growth\nFY2025 Q3 · YoY"),
    (f"{latest:,}","UK sample vacancies\nSix months to 19 Oct 2026"),
    (f"+{share_delta:.2f} pp","UK vacancy share change\n0.21% → 0.22%")]):
    ax=fig.add_subplot(gs[0,i]); ax.axis("off")
    ax.text(0,.65,value,fontsize=29,fontweight="bold",color=TEAL,transform=ax.transAxes)
    ax.text(0,.03,label,fontsize=10,color="#596779",transform=ax.transAxes)
ax=fig.add_subplot(gs[1,:2]); bars=ax.bar(hiring.Period.astype(str),hiring.Vacancies,color=[BLUE,TEAL],width=.52)
ax.bar_label(bars,padding=6,fontsize=14); ax.set_ylim(0,310)
ax.set_ylabel("Vacancies in sample"); ax.set_title("UK vacancy counts: two sample windows",loc="left",pad=16)
ax=fig.add_subplot(gs[1,2:]); ax.axis("off")
ax.text(0,.95,"The distinction that matters",fontsize=17,fontweight="bold",va="top")
ax.text(0,.72,f"Count change: +{growth:.1f}%\nShare change: +{share_delta:.2f} percentage points\n\nA larger raw vacancy count does not establish\nan equally large rise in relative demand.\n\n2027 outlook: qualitative only.\nNo numeric industry forecast is supplied.",fontsize=12,linespacing=1.65,va="top")
ax=fig.add_subplot(gs[2,:]); ax.axis("off")
ax.text(0,.5,"SUPPLIED SNAPSHOT • Market claims unverified • Microsoft / IT Jobs Watch attribution in supplied text; original citations absent",fontsize=9,color="#8A5730")
fig.suptitle("POWER PLATFORM | Opportunity with evidence in context",fontsize=20,fontweight="bold",x=.08,ha="left",y=1.02)
plt.show()
```

## 3 · Salary explorer
Each country uses its own currency, axis and pay period. These panels do **not** compare purchasing power or savings. Ranges also have different statistical meanings. Guide ranges are not overseas career-switch offers; UAE examples are closed ads.

The India negotiation estimates appear separately below the benchmark panels. **Base pay and CTC are different.**


```python
# @title View the supplied country salary benchmarks
salary=frames["Salary"]
benchmarks=salary[salary.EvidenceType.ne("Advisory estimate")]
fig, axes=plt.subplots(4,2,figsize=(13,13))
for ax,country in zip(axes.flat,["India","UK","Germany","Australia","Canada","US","UAE"]):
    sub=benchmarks[benchmarks.Country.eq(country)].reset_index(drop=True)
    for i,row in sub.iterrows():
        ax.plot([row.Low,row.High],[i,i],color=TEAL,lw=7,solid_capstyle="round")
        if pd.notna(row.Central): ax.scatter(row.Central,i,color=AMBER,s=65,zorder=3)
        ax.text(row.Low,i+.13,f"{row.Low:,.1f}" if country=="India" else f"{row.Low:,.0f}",fontsize=10)
        ax.text(row.High,i+.13,f"{row.High:,.1f}" if country=="India" else f"{row.High:,.0f}",ha="right",fontsize=10)
    ax.set_title(f"{country} · {sub.Currency.iloc[0]} · {sub.Period.iloc[0]}",loc="left",fontsize=12)
    ax.set_xlim(0,float(sub.High.max())*1.2); ax.set_ylim(-.45,len(sub)-.15)
    ax.set_yticks(range(len(sub)),["Benchmark" if len(sub)==1 else f"Closed ad {i+1}" for i in range(len(sub))])
    ax.set_xlabel(sub.Statistic.iloc[0],fontsize=9)
    ax.xaxis.set_major_formatter(ticker.StrMethodFormatter("{x:,.0f}"))
axes.flat[-1].axis("off"); axes.flat[-1].text(0,.9,"READ BEFORE COMPARING\n\nIndependent scales and currencies.\nAmber dot = reported mean / median.\nNo currency conversion or salary average.\nAll benchmarks require source verification.",va="top",fontsize=12,linespacing=1.8)
fig.suptitle("SALARY BENCHMARKS | Supplied research, distinct contexts",fontsize=18,fontweight="bold",y=1.01)
fig.tight_layout(h_pad=3,w_pad=3); plt.show()
show_table(benchmarks[["Country","Currency","Period","Low","High","Central","Statistic","SourceClaim","Caveat"]])
```

```python
# @title India negotiation estimates — advisory, not observed market pay
estimates=salary[salary.EvidenceType.eq("Advisory estimate")].reset_index(drop=True)
fig,ax=plt.subplots(figsize=(12,4.5))
for i,row in estimates.iterrows():
    ax.plot([row.Low,row.High],[i,i],color=AMBER,lw=9,solid_capstyle="round")
    upper="22+" if "Established" in row.Role else f"{row.High:g}"
    ax.text(row.Low,i+.17,f"{row.Low:g}–{upper}",fontsize=11)
ax.set_yticks(range(len(estimates)),estimates.Role); ax.invert_yaxis()
ax.set_xlim(0,26); ax.set_xlabel("INR lakh / year · CTC")
ax.set_title("Illustrative application ranges by evidence of delivery",loc="left")
fig.text(.08,-.04,"Advisory estimates only. The specialist 22+ endpoint is open-ended; the drawn line ends at 22 for display.",fontsize=10,color="#8A5730")
fig.tight_layout(); plt.show()
```

```python
# @title Edit the illustrative salary scenario and rerun this cell
current_lpa = 7.0 # @param {type:"number"}
target_lpa = 15.0 # @param {type:"number"}
if current_lpa <= 0 or target_lpa < 0:
    raise ValueError("Current pay must be positive and target pay cannot be negative.")
increase=(target_lpa/current_lpa-1)*100
fig,ax=plt.subplots(figsize=(8,3.2))
bars=ax.barh(["Current illustration","Target illustration"],[current_lpa,target_lpa],color=[BLUE,TEAL])
ax.bar_label(bars,labels=[f"{current_lpa:g} LPA",f"{target_lpa:g} LPA"],padding=5)
ax.set_xlim(0,max(current_lpa,target_lpa)*1.25)
ax.set_xlabel("INR lakh / year · compare like-for-like pay components")
ax.set_title(f"Editable scenario: {increase:+.1f}% change",loc="left")
fig.tight_layout(); plt.show()
print("Illustrative inputs only; this is neither a verified personal salary nor a hiring prediction.")
```

Tables are intentionally disconnected because their grains differ. Joining salary ranges, adoption statistics, role advice and roadmap stages would create misleading relationships. No cross-currency salary total is calculated.


```python
# @title Inspect all table grains and row counts
schema=pd.DataFrame([{"Table":name,"Rows":len(frame),"Columns":len(frame.columns),"Grain":{
"Adoption":"One metric / reporting period", "UKHiring":"One UK six-month sample / ending year",
"Salary":"One benchmark or advisory estimate", "Roles":"One target role", "Roadmap":"One two-week stage",
"Outlook":"One qualitative opportunity area", "Evidence":"One evidence limitation", "Scenario":"One editable illustrative scenario"}[name]}
 for name,frame in frames.items()])
show_table(schema)
assert sum(schema.Rows)==43
assert salary.BenchmarkID.is_unique
assert abs(growth-151.92307692307693)<1e-8
assert abs(share_delta-.01)<1e-8
print("PASS: 43 rows; unique salary benchmark IDs; hiring-count and percentage-point arithmetic.")
```
